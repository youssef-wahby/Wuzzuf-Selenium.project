## selenium project for job info

```python
# our objective is to utilize selenium using ython to collect and scrape hiring infro from wuzzuf.com 
```

```python
#1-lets start by importing our librares that we will need to make this project.
import selenium # this is our main library that we are going to use for automation and web scrapping
import time # we will use the time library to allow smooth transitions between the website pages 
from selenium import webdriver
from selenium.webdriver.common.by import By 
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager # we will be using chrome for our project 
```

```python
#2-lets save our url in a summerized way to keep a simplistic look.
URL= "https://wuzzuf.net/"
```

```python
#3- we are going to be using these options just to avoid the bot detection system.
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new") # we can use this at the very end to make our process seamless, lets save that for later
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
```

```python
# we are going to be searching for info about the following jobs:
#Data Engineer
#Data Science 
#AI Developer
#Machine Learning Engineer
```

```python
#4- lets start making selenium open our website. 
driver = webdriver.Chrome()
driver.get(URL)
time.sleep(10)#--> to allow our website some time to load 
driver.quit()
```

```python
#5-now lets make selenium start inputing our search data and have it click search.
driver = webdriver.Chrome()
driver.get(URL)
time.sleep(1)
driver.find_element(By.NAME,"q").send_keys("Data Engineer")#--> we here used the name of the search bar to lock onto it and input out first job
time.sleep(1)
driver.find_element(By.CLASS_NAME,"css-1stvrxn").click()#--> we found the class name for the search button
time.sleep(2)
driver.refresh()#--> we are going to be refreshing to avoid the app interface blocking selenium
time.sleep(10)
driver.quit()
```

```python
#6- here we are just going to be making a list to store our data temporarily 
list_of_job_info=[]
```

```python
#7-here we are going to start collecting data from the first page of our first job 
driver = webdriver.Chrome()
driver.get(URL)
time.sleep(1)
driver.find_element(By.NAME,"q").send_keys("Data Engineer")
time.sleep(1)
driver.find_element(By.CLASS_NAME,"css-1stvrxn").click()
time.sleep(2)
driver.refresh()
time.sleep(10)
###-----------lets make a small loop that goes over every job
jobs = driver.find_elements(By.CLASS_NAME,"css-ghe2tq")#class name for every job card 
for job in jobs:
    Job_title = job.find_element(By.CLASS_NAME,"css-o171kl").text
    Company=job.find_element(By.CLASS_NAME,"css-ipsyv7").text
    Location=job.find_element(By.CLASS_NAME,"css-16x61xq").text
    Job_Type=job.find_element(By.CLASS_NAME,"css-uc9rga").text
    Job_URL=job.find_element(By.CLASS_NAME,"css-o171kl").get_attribute("href")#--> since link doesnt have class we can use its attribute  

    list_of_job_info.append({
        "Job_title":Job_title,
        "Company":Company,
        "Location":Location,
        "Job_Type":Job_Type,
        "Job_URL":Job_URL
    })
time.sleep(10)
driver.quit()
```

```python
#8- now we are going to make selenium go over multiple pages 
driver = webdriver.Chrome()
driver.get(URL)
time.sleep(1)
driver.find_element(By.NAME,"q").send_keys("Data Engineer")
time.sleep(1)
driver.find_element(By.CLASS_NAME,"css-1stvrxn").click()
time.sleep(2)
driver.refresh()
time.sleep(10)
list_of_jobs = []
for i in range (2):
  jobs = driver.find_elements(By.CLASS_NAME,"css-ghe2tq")#outer loop to check first 3 pages 
  time.sleep(5)
  for job in jobs:
     Job_title = job.find_element(By.CLASS_NAME,"css-o171kl").text
     Company=job.find_element(By.CLASS_NAME,"css-ipsyv7").text
     Location=job.find_element(By.CLASS_NAME,"css-16x61xq").text
     Job_Type=job.find_element(By.CLASS_NAME,"css-uc9rga").text
     Job_URL=job.find_element(By.CLASS_NAME,"css-o171kl").get_attribute("href")

     list_of_job_info.append({
        "Job_title":Job_title,
        "Company":Company,
        "Location":Location,
        "Job_Type":Job_Type,
        "Job_URL":Job_URL
      })
  try:
     driver.find_element(By.CSS_SELECTOR, "button[class='css-kefogq ezfki8j0']").click()#clicking on next page button 
     time.sleep(3)
  except:
      break


time.sleep(10)
driver.quit()
```

```python
#9-now lets loop around multiple job searches
import pandas as pd 
positions = ['Ai Developer', 'Data Analyst', 'Data Scientist', 'Data Engineer'] 
 
options = Options() 
options.add_argument("--disable-blink-features=AutomationControlled") 
options.add_experimental_option("excludeSwitches", ["enable-automation"]) 
options.add_experimental_option('useAutomationExtension', False) 
#options.add_argument("--headless=new") # we can use this at the very end to make our process seamless, lets save that for later 
options.add_argument("--window-size=1920,1080") 
options.add_argument("--disable-gpu") 
options.add_argument("--no-sandbox") 
options.add_argument("--disable-dev-shm-usage") 
driver = webdriver.Chrome(options=options) 
driver.get(URL) 
list_of_job_info= [] 

for position in positions: 

    time.sleep(2) 
    driver.find_element(By.NAME,"q").send_keys(position) 
 
    driver.find_element(By.XPATH, "//button[@type='submit']").click() 
 
    time.sleep(2) 
 
    driver.refresh() 
 
    time.sleep(7) 
 
    for i in range(5): 

        time.sleep(3) 
        jobs = driver.find_elements(By.CLASS_NAME,"css-ghe2tq") # outer loop to check first 5 pages
        time.sleep(1) 

        for job in jobs: 

            Job_title = job.find_element(By.CLASS_NAME,"css-o171kl").text 
            Company = job.find_element(By.CLASS_NAME,"css-ipsyv7").text 
            Location = job.find_element(By.CLASS_NAME,"css-16x61xq").text 
            Job_Type = job.find_element(By.CLASS_NAME,"css-uc9rga").text 
            Job_URL = job.find_element(By.CLASS_NAME,"css-o171kl").get_attribute("href"), 

            try: 
                job_exp = job.find_element(
                    By.CSS_SELECTOR, 
                    "div > div.css-1rhj4yg > div:nth-child(2) > span"
                ).text 

            except: 
                job_exp = '' ,

            job_occupation = job.find_element(By.CLASS_NAME,"css-uofntu").text 
 
            list_of_job_info.append({ 
                "Job_title":Job_title, 
                "Company":Company, 
                "Location":Location, 
                "Job_Type":Job_Type, 
                "Job_URL":Job_URL, 
                "Job Experience":job_exp, 
                "job occupation":job_occupation  
            }) 

            time.sleep(1) 

        # NEXT PAGE happens AFTER collecting all jobs on the current page
        try: 
                           next_button = driver.find_element(
                               By.XPATH, "(//*[contains(@class, 'css-kefogq')])[last()]"# here we can observe some new syntax since we have two buttons with the same class we need to have a work around for it 
                           )# we are just finding the element and adding last to it bec its the next button thatbwe wanna Click  on and not the back button 
                           next_button.click()
                           time.sleep(5)  
        except: 
                break 
 
df = pd.DataFrame(list_of_job_info)
```
